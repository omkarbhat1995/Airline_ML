# Data Transformation

```python
import joblib
import pandas as pd
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
```

```python
X_train = pd.read_csv('G:\\Ethans\\ML\\Projects\\Final Projects\\Airline_ML\\Data\\Cleaned_Split_Data\\X_train.csv')
X_test = pd.read_csv('G:\\Ethans\\ML\\Projects\\Final Projects\\Airline_ML\\Data\\Cleaned_Split_Data\\X_test.csv')
y_train = pd.read_csv('G:\\Ethans\\ML\\Projects\\Final Projects\\Airline_ML\\Data\\Cleaned_Split_Data\\y_train.csv')
y_test = pd.read_csv('G:\\Ethans\\ML\\Projects\\Final Projects\\Airline_ML\\Data\\Cleaned_Split_Data\\y_test.csv')
```

## Creating the Transformer

```python
# Transforming encoding categorical variables into numerical values using one-hot encoding
numerical_columns = ['Total_Stops', 'Duration_Hours_Minutes', 'Day_of_Departure', 'Month_of_Departure', 'Year_of_Departure', 'Dep_Time_Hour', 'Dep_Time_Minute', 'Arrival_Time_Hour', 'Arrival_Time_Minute', 'Year_of_Arrival', 'Month_of_Arrival', 'Day_of_Arrival']
one_hot_columns = ['Airline', 'Source', 'Destination']



OneHotEncoding_01 =  ("ohe",OneHotEncoder(sparse_output=False, drop="first", handle_unknown="ignore"),one_hot_columns,)

Scaling = ("num", StandardScaler(), numerical_columns)
transform_engine = ColumnTransformer(
            transformers=[OneHotEncoding_01,Scaling,]).set_output(transform="pandas")
```

```python
joblib.dump(transform_engine, "G:\\Ethans\\ML\\Projects\\Final Projects\\Airline_ML\\Transformation_files\\transform.pkl")
print("transform.pkl generated successfully!")
```

## Transforming the Data

```python
transform_engine = joblib.load("G:\\Ethans\\ML\\Projects\\Final Projects\\Airline_ML\\Transformation_files\\transform.pkl")
```

```python
X_train_transformed = transform_engine.fit_transform(X_train)
X_test_transformed = transform_engine.transform(X_test)
```
