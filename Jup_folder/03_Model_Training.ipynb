# Model Training

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import joblib
```

```python
# Reading the transformed training and testing datasets and their corresponding target variables
X_train_transformed = pd.read_csv('G:\\Ethans\\ML\\Projects\\Final Projects\\Airline_ML\\Data\\Transformed_Data\\X_train_transformed.csv')
y_train = pd.read_csv('G:\\Ethans\\ML\\Projects\\Final Projects\\Airline_ML\\Data\\Cleaned_Split_Data\\y_train.csv')
X_test_transformed = pd.read_csv('G:\\Ethans\\ML\\Projects\\Final Projects\\Airline_ML\\Data\\Transformed_Data\\X_test_transformed.csv')
y_test = pd.read_csv('G:\\Ethans\\ML\\Projects\\Final Projects\\Airline_ML\\Data\\Cleaned_Split_Data\\y_test.csv')
```

## Selecting the model and model training

### Linear Model

```python
model = LinearRegression()
model.fit(X_train_transformed, y_train)
```

```python
joblib.dump(model, "G://Ethans//ML//Projects//Final Projects//Airline_ML//Models//linear_model.pkl")
print("Saved linear_model.pkl successfully!")
```

### Lasso Model

```python
from sklearn.linear_model import Lasso

lasso_model = Lasso(alpha=0.1)
lasso_model.fit(X_train_transformed, y_train)
joblib.dump(lasso_model, 'G://Ethans//ML//Projects//Final Projects//Airline_ML//Models//lasso_model.pkl')
```

```python
from sklearn.linear_model import Ridge
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(X_train_transformed, y_train)
joblib.dump(ridge_model, 'G://Ethans//ML//Projects//Final Projects//Airline_ML//Models//ridge_model.pkl')
```

## Evaluating the Model

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
def linear_metrics(y_true, y_pred, X):
    MSE = mean_squared_error(y_true, y_pred)
    RMSE = np.sqrt(MSE)
    MAE = mean_absolute_error(y_true, y_pred)
    R2 = r2_score(y_true, y_pred)
    Adg_R2 = 1 - (1-R2)*(len(y_true)-1)/(len(y_true)-X.shape[1]-1)
    return {"MSE": MSE, "RMSE": RMSE, "MAE": MAE, "R2": R2, "Adjusted R2": Adg_R2}
```

### Linear model

```python
linear_metrics(y_train, y_train_pred_linear, X_train_transformed)
```

```python
linear_metrics(y_test, y_test_pred_linear, X_test_transformed)
```

### Ridge Model

```python
linear_metrics(y_train, y_train_pred_ridge, X_train_transformed)
```

```python
linear_metrics(y_test, y_test_pred_ridge, X_test_transformed)
```

### Lasso Model

```python
linear_metrics(y_train, y_train_pred_lasso, X_train_transformed)
```

```python
linear_metrics(y_test, y_test_pred_ridge, X_test_transformed)
```

### Elastic Net Model

```python
linear_metrics(y_train, y_train_pred_elastic_net, X_train_transformed)
```

```python
linear_metrics(y_test, y_test_pred_elastic_net, X_test_transformed)
```
